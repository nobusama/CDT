# CDT Paper - Figure 7: Two Windows Conceptual Diagram

This figure illustrates the "Two Windows" framework:
- **Attention**: What the model "looks at" during inference
- **Gradient**: What "drives" the prediction

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch, Circle, Rectangle
from matplotlib.lines import Line2D
import matplotlib.patheffects as path_effects

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['font.size'] = 11
```

```python
fig = plt.figure(figsize=(14, 8), dpi=300)

# Main axis
ax = fig.add_subplot(111)
ax.set_xlim(0, 14)
ax.set_ylim(0, 8)
ax.axis('off')

# Title
ax.text(7, 7.7, 'Two Windows into Regulatory Mechanism', 
        ha='center', va='top', fontsize=16, fontweight='bold')

# ============ LEFT SIDE: ATTENTION ============
# Box for Attention
ax.add_patch(FancyBboxPatch((0.5, 2.5), 5.5, 4.5, 
                             boxstyle="round,pad=0.1", 
                             facecolor='#E3F2FD', edgecolor='#1976D2', 
                             linewidth=2))

ax.text(3.25, 6.7, 'ATTENTION', ha='center', va='bottom', 
        fontsize=14, fontweight='bold', color='#1976D2')
ax.text(3.25, 6.3, '"What the model looks at"', ha='center', va='bottom', 
        fontsize=10, style='italic', color='#1976D2')

# Eye icon (simplified)
eye_x, eye_y = 3.25, 5.3
ax.add_patch(Circle((eye_x, eye_y), 0.4, facecolor='#1976D2', edgecolor='white', linewidth=2))
ax.add_patch(Circle((eye_x, eye_y), 0.15, facecolor='white'))
ax.plot([eye_x-0.6, eye_x+0.6], [eye_y, eye_y], 'k-', lw=2)
ax.plot([eye_x-0.5, eye_x-0.7], [eye_y+0.2, eye_y+0.1], 'k-', lw=2)
ax.plot([eye_x+0.5, eye_x+0.7], [eye_y+0.2, eye_y+0.1], 'k-', lw=2)

# Attention description
attention_points = [
    'Spatial patterns learned',
    'Features incorporated during inference',
    'May be diffuse (uniform attention)',
    'Shows model\'s "field of view"'
]
for i, point in enumerate(attention_points):
    ax.text(1.0, 4.5 - i*0.5, f'• {point}', ha='left', va='center', fontsize=9, color='#0D47A1')

# ============ RIGHT SIDE: GRADIENT ============
# Box for Gradient
ax.add_patch(FancyBboxPatch((8, 2.5), 5.5, 4.5, 
                             boxstyle="round,pad=0.1", 
                             facecolor='#FFEBEE', edgecolor='#C62828', 
                             linewidth=2))

ax.text(10.75, 6.7, 'GRADIENT', ha='center', va='bottom', 
        fontsize=14, fontweight='bold', color='#C62828')
ax.text(10.75, 6.3, '"What drives the prediction"', ha='center', va='bottom', 
        fontsize=10, style='italic', color='#C62828')

# Target icon (simplified)
target_x, target_y = 10.75, 5.3
ax.add_patch(Circle((target_x, target_y), 0.45, facecolor='white', edgecolor='#C62828', linewidth=2))
ax.add_patch(Circle((target_x, target_y), 0.3, facecolor='white', edgecolor='#C62828', linewidth=2))
ax.add_patch(Circle((target_x, target_y), 0.15, facecolor='#C62828', edgecolor='#C62828'))

# Gradient description
gradient_points = [
    'Input features critical for output',
    'Causal influence on prediction',
    'Sharp peaks at functional elements',
    'Reveals prediction-critical positions'
]
for i, point in enumerate(gradient_points):
    ax.text(8.5, 4.5 - i*0.5, f'• {point}', ha='left', va='center', fontsize=9, color='#B71C1C')

# ============ CENTER: Comparison ============
# Arrow showing "different but complementary"
ax.annotate('', xy=(7.8, 4.75), xytext=(6.2, 4.75),
            arrowprops=dict(arrowstyle='<->', color='#424242', lw=2))
ax.text(7, 5.1, '~10% overlap', ha='center', va='bottom', fontsize=10, 
        fontweight='bold', color='#424242')
ax.text(7, 4.4, 'Complementary', ha='center', va='top', fontsize=9, 
        style='italic', color='#616161')

# ============ BOTTOM: Example (FNDC5) ============
ax.add_patch(FancyBboxPatch((1, 0.3), 12, 1.8, 
                             boxstyle="round,pad=0.05", 
                             facecolor='#F5F5F5', edgecolor='#9E9E9E', 
                             linewidth=1.5))

ax.text(7, 1.9, 'FNDC5 Case Study Example', ha='center', va='bottom', 
        fontsize=11, fontweight='bold', color='#424242')

# Left: Attention result
ax.add_patch(FancyBboxPatch((1.5, 0.6), 4.5, 1.0, 
                             boxstyle="round,pad=0.02", 
                             facecolor='#BBDEFB', edgecolor='#1976D2', 
                             linewidth=1))
ax.text(3.75, 1.35, 'Attention peak: +47.4 kb', ha='center', va='center', 
        fontsize=10, fontweight='bold', color='#1976D2')
ax.text(3.75, 0.9, 'No clear biological annotation', ha='center', va='center', 
        fontsize=9, color='#1565C0')

# Right: Gradient result
ax.add_patch(FancyBboxPatch((8, 0.6), 4.5, 1.0, 
                             boxstyle="round,pad=0.02", 
                             facecolor='#FFCDD2', edgecolor='#C62828', 
                             linewidth=1))
ax.text(10.25, 1.35, 'Gradient peak: -56.7 kb', ha='center', va='center', 
        fontsize=10, fontweight='bold', color='#C62828')
ax.text(10.25, 0.9, 'CTCF binding site → Hi-C validated', ha='center', va='center', 
        fontsize=9, color='#B71C1C')

# Arrow between examples
ax.annotate('', xy=(7.8, 1.1), xytext=(6.2, 1.1),
            arrowprops=dict(arrowstyle='<->', color='#757575', lw=1.5))
ax.text(7, 1.1, '≠', ha='center', va='center', fontsize=14, fontweight='bold', color='#424242')

plt.tight_layout()
plt.savefig('figure7_two_windows.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

print("\nFigure 7 saved as 'figure7_two_windows.png'")
```
